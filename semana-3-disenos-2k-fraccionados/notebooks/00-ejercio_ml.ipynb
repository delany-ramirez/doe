```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import statsmodels.formula.api as smf
import statsmodels.api as sm

np.random.seed(3)
```

```python
df = pd.read_csv('../datos/datos_ml.csv', delimiter=';', decimal=',')
df
```

```python
# Entrenar un modelo de regresión lineal usando los datos de df con los factores A, B y C y sus interacciones como predictores y la variable de respuesta Accuracy.
df['Accuracy'] = pd.to_numeric(
    df['Accuracy'].astype(str).str.replace(',', '.', regex=False),
    errors='coerce'
)
df[['A', 'B', 'C']] = df[['A', 'B', 'C']].apply(pd.to_numeric, errors='coerce')

modelo = smf.ols('Accuracy ~ A*B*C', data=df).fit()
```

```python
# Imprimir el resumen del modelo
print(modelo.summary())
```
