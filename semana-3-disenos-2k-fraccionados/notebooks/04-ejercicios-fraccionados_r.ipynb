## Ejercicio 1 · Helicóptero de papel ($2^{5-2}$, Resolución III)

8 corridas, 5 factores. Generadores: $D=AB$, $E=AC$.  
Relacion de definicion: $I = ABD = ACE = BCDE$ — **Resolución III**.  
Alias clave: $[A]=A+BD+CE$, $[B]=B+AD$, $[C]=C+AE$, $[BC]=BC+DE$.

```r
df1 <- read.csv('../datos/papel-helicoptero-frac.csv')
cat('Diseno 2^{5-2} — 8 corridas:\n')
print(df1)

# Verificar generadores D=AB, E=AC
cat('\nVerificacion D=AB:', all(df1$tipo_papel == df1$largo_ala * df1$ancho_ala), '\n')
cat('Verificacion E=AC:', all(df1$clip == df1$largo_ala * df1$largo_cuerpo), '\n')
```

```r
# Efectos principales (7 df con 8 corridas)
modelo1 <- lm(tiempo_vuelo ~ largo_ala + ancho_ala + largo_cuerpo + tipo_papel + clip,
              data = df1)
efectos1 <- coef(modelo1)[-1] * 2
names(efectos1) <- c('A','B','C','D','E')
cat('Efectos estimados (cada uno confundido con 2FIs en Res. III):\n')
print(round(efectos1, 3))
```

```r
# Grafica normal de efectos
eff1 <- sort(efectos1)
n1 <- length(eff1)
z1 <- qnorm((seq_len(n1) - 0.5) / n1)

plot(z1, eff1, pch = 16, col = 'steelblue', cex = 1.4,
     xlab = 'Cuantiles normales', ylab = 'Efecto estimado (s)',
     main = 'Grafica normal de efectos — Helicoptero de papel')
abline(h = 0, lty = 2, col = 'gray')
text(z1, eff1, names(eff1), pos = 4, cex = 1.0)
grid()
cat('Factores activos: A (largo_ala) y B (ancho_ala).\n')
cat('Interpretar con cautela: [A]=A+BD+CE en Res. III.\n')
```

## Ejercicio 2 · Cribado proceso quimico ($2^{6-2}$, Resolución IV)

16 corridas, 6 factores. Generadores: $E=ABC$, $F=BCD$.  
Relacion de definicion: $I=ABCE=BCDF=ADEF$ — **Resolución IV**.  
Efectos principales limpios; 2FIs aliadas entre si.

```r
df2 <- read.csv('../datos/proceso-quimico-frac.csv')
cat('Diseno 2^{6-2} — 16 corridas:\n')
print(df2)

# Verificar generadores
cat('\nVerificacion E=ABC:', all(df2$E == df2$A * df2$B * df2$C), '\n')
cat('Verificacion F=BCD:', all(df2$F == df2$B * df2$C * df2$D), '\n')
```

```r
# Efectos principales (limpios en Res. IV)
modelo2 <- lm(rendimiento ~ A + B + C + D + E + F, data = df2)
efectos2 <- coef(modelo2)[-1] * 2
cat('Efectos principales estimados (%):\n')
print(round(efectos2, 2))
cat('R2 solo efectos principales:', summary(modelo2)$r.squared, '\n')
```

```r
eff2 <- sort(efectos2)
n2 <- length(eff2)
z2 <- qnorm((seq_len(n2) - 0.5) / n2)

plot(z2, eff2, pch = 16, col = 'darkgreen', cex = 1.4,
     xlab = 'Cuantiles normales', ylab = 'Efecto estimado (%)',
     main = 'Grafica normal — Proceso quimico (6 factores, 16 corridas)')
abline(h = 0, lty = 2, col = 'gray')
text(z2, eff2, names(eff2), pos = 4, cex = 1.0)
grid()
```

```r
# Modelo reducido con factores activos
modelo2r <- lm(rendimiento ~ A + C + D, data = df2)
print(summary(modelo2r))
cat('\nNOTA: En Res. IV, 2FIs como AC y CD estan aliadas entre si.\n')
cat('Si se sospecha interaccion, hacer fold-over para desconfundir.\n')
```

## Ejercicio 3 · Tratamiento de agua ($2^{4-1}$, Resolución IV)

8 corridas, 4 factores. Generador: $D=ABC$ ($I=ABCD$, **Resolución IV**).  
Alias: $[A]=A+BCD$, $[B]=B+ACD$, $[C]=C+ABD$, $[D]=D+ABC$,  
$[AB]=AB+CD$, $[AC]=AC+BD$, $[AD]=AD+BC$.

```r
df3 <- read.csv('../datos/tratamiento-agua-frac.csv')
cat('Diseno 2^{4-1} — 8 corridas:\n')
print(df3)

# Verificar D=ABC
cat('\nVerificacion D=ABC:', all(df3$dosis_cloro == df3$pH * df3$temperatura * df3$tiempo_contacto), '\n')
```

```r
# Efectos principales (limpios de 2FIs en Res. IV)
modelo3 <- lm(turbidez ~ pH + temperatura + tiempo_contacto + dosis_cloro, data = df3)
efectos3 <- coef(modelo3)[-1] * 2
names(efectos3) <- c('A(pH)','B(temp)','C(tiempo)','D(cloro)')
cat('Efectos principales estimados (NTU):\n')
print(round(sort(efectos3), 3))

eff3 <- sort(efectos3)
n3 <- length(eff3)
z3 <- qnorm((seq_len(n3) - 0.5) / n3)

plot(z3, eff3, pch = 16, col = 'navy', cex = 1.6,
     xlab = 'Cuantiles normales', ylab = 'Efecto (NTU)',
     main = 'Grafica normal — Tratamiento de agua')
abline(h = 0, lty = 2, col = 'gray')
text(z3, eff3, names(eff3), pos = 4, cex = 0.95)
grid()
cat('\nFactores dominantes: D(dosis_cloro) negativo y B(temperatura) positivo.\n')
cat('Objetivo: minimizar turbidez -> alta dosis_cloro, alto pH, baja temperatura.\n')
```

```r
# Aliases 2FIs en I=ABCD
cat('Aliased 2FIs (I=ABCD):\n')
alias2fi <- data.frame(estimado = c('AB','AC','AD'), aliasado = c('CD','BD','BC'))
print(alias2fi)
cat('\nPara separar AB de CD: fold-over (16 corridas = 2^4 completo).\n')

# Prediccion optima
opt3 <- data.frame(pH = 1, temperatura = -1, tiempo_contacto = 0, dosis_cloro = 1)
cat('\nPrediccion (pH=+1, temp=-1, DC=+1):', predict(modelo3, opt3), 'NTU\n')
```

## Ejercicio 4 · Biofermentación ($2^{5-1}$, Resolución V)

16 corridas, 5 factores. Generador: $E=ABCD$ ($I=ABCDE$, **Resolución V**).  
Con Resolucion V: **efectos principales y todas las 2FIs son estimables limpiamente.**

| Factor | $-1$ | $+1$ |
|--------|------|------|
| pH (A) | 6.0 | 7.5 |
| temperatura (B) | 28 °C | 37 °C |
| oxigeno (C) | 20 % | 40 % |
| sustrato (D) | 10 g/L | 20 g/L |
| agitacion (E=ABCD) | 100 rpm | 200 rpm |

```r
df4 <- read.csv('../datos/biofermentacion-frac.csv')
cat('Diseno 2^{5-1} Res. V — 16 corridas:\n')
print(df4)

# Verificar E = A*B*C*D
cat('\nVerificacion E=ABCD:', all(df4$agitacion == df4$pH * df4$temperatura * df4$oxigeno * df4$sustrato), '\n')
```

```r
# Modelo con efectos principales y TODAS las 2FIs (estimables en Res. V)
modelo4 <- lm(enzima ~ (pH + temperatura + oxigeno + sustrato + agitacion)^2, data = df4)
efectos4 <- coef(modelo4)[-1] * 2
nombres4 <- c('A','B','C','D','E','AB','AC','AD','AE','BC','BD','BE','CD','CE','DE')
names(efectos4) <- nombres4
cat('Efectos estimados (U/mL — por magnitud):\n')
print(round(sort(abs(efectos4), decreasing = TRUE), 2))
```

```r
eff4 <- sort(efectos4)
n4 <- length(eff4)
z4 <- qnorm((seq_len(n4) - 0.5) / n4)

plot(z4, eff4, pch = 16, col = 'purple', cex = 1.2,
     xlab = 'Cuantiles normales', ylab = 'Efecto estimado (U/mL)',
     main = 'Grafica normal de efectos — Biofermentacion (Res. V, 16 corridas)')
abline(h = 0, lty = 2, col = 'gray')
text(z4, eff4, names(eff4), pos = 4, cex = 0.85)
grid()
```

```r
# Modelo reducido: A, B, D, AB (efectos activos)
modelo4r <- lm(enzima ~ pH + temperatura + sustrato + pH:temperatura, data = df4)
print(summary(modelo4r))

op <- par(mfrow = c(1, 2))
qqnorm(residuals(modelo4r), main = 'Q-Q residuales - biofermentacion')
qqline(residuals(modelo4r))
plot(fitted(modelo4r), residuals(modelo4r),
     xlab = 'Ajustados (U/mL)', ylab = 'Residuales',
     main = 'Residuales vs. ajustados')
abline(h = 0, lty = 2)
par(op)
print(shapiro.test(residuals(modelo4r)))

cat('\nProyeccion: un 2^{5-1} Res. V se proyecta en cualquier conjunto de 4 factores\n')
cat('como un factorial completo 2^4.\n')
```
